```python
import pandas as pd
import numpy as np

# Import CSV using Pandas
df = pd.read_csv("social_media_engagement_5000.csv")
#Pandas is the standard Python tool for data analysis. It lets Python read, filter, and change data exactly like an Excel spreadsheet or a relational database table.

# Check data types before conversion
print("Original Data Types:")
print(df.dtypes)
#This reads your CSV file and saves it as a DataFrame (named df). A DataFrame is simply a grid of rows and columns in the computer's memory.

# Convert date column to datetime (replace 'date_column_name' with your actual column name)
# Example: df['post_date'] = pd.to_datetime(df['post_date'])

# Demonstrate using a NumPy array for basic array operations
likes_array = np.array(df['likes'].dropna())
print(f"Max likes using NumPy: {np.max(likes_array)}")
#NumPy arrays are highly optimized lists. They are used here because they perform mathematical calculations much faster than standard Python lists.
```

```python
# Detect missing values
print("Missing values per column:\n", df.isnull().sum())
#This scans the entire table and counts exactly how many empty cells exist in each column.
```

```python
# Handle missing data (Filling numeric columns with median, categorical with mode)
# Adjust column names as they appear in your dataset
df['likes'] = df['likes'].fillna(df['likes'].median())
df['comments'] = df['comments'].fillna(df['comments'].median())
#Blank cells crash mathematical equations. This code finds all missing values in the 'likes' column and replaces them with a safe fallback number, like the median.
```

```python
# Identify and remove duplicates
duplicates_count = df.duplicated().sum()
print(f"Found {duplicates_count} duplicates. Removing them...")
df = df.drop_duplicates()
# This removes any exact copycat rows so your final counts are not artificially inflated.
```

```python
# Standardize categories (e.g., converting gender labels to title case)
if 'gender' in df.columns:
    df['gender'] = df['gender'].str.title()

```

```python
#Extract hashtag count (Assuming hashtags are in a 'text' or 'content' column)
if 'content' in df.columns:
    df['hashtag_count'] = df['content'].apply(lambda x: str(x).count('#'))
    #A lambda is a quick, one-line function. This specific line goes down every row in the text column, counts the # symbol, and saves that number in a new column.
```

```python
# View dataset structure
print("First 5 rows:\n", df.head())
print("\nDataset shape:", df.shape)
#This prints only the first 5 rows of your table. It is a quick visual check to ensure the data loaded correctly without flooding your screen with thousands of rows.
```

```python
# Generate summary statistics
print("\nSummary Statistics:\n", df.describe())
```

```python
# Analyze categorical distributions
if 'post_type' in df.columns:
    print("\nPost Type Distribution:\n", df['post_type'].value_counts())
    #Returns two numbers: the total count of rows and the total count of columns.
```

```python
# Create a correlation matrix for numeric fields
# Select only numeric columns first to avoid errors
numeric_df = df.select_dtypes(include=[np.number])
print("\nCorrelation Matrix:\n", numeric_df.corr())
#
```

```python
# Summarize metrics using groupby (e.g., avg likes by post type)
if 'post_type' in df.columns and 'likes' in df.columns:
    print("\nAvg Likes by Post Type:\n", df.groupby('post_type')['likes'].mean())
    # This works identically to a GROUP BY statement in SQL. It groups all identical post types (like grouping all "Video" rows together) and calculates the average likes for that specific group.
```

```python
# Create new field: engagement_score (likes + comments + shares)
# Ensure columns exist before adding
if all(col in df.columns for col in ['likes', 'comments', 'shares']):
    df['engagement_score'] = df['likes'] + df['comments'] + df['shares']
#This creates a brand new column named 'engagement_score' on the far right of your table by simply adding the numbers in the existing columns together row by row.
```

```python
# Create an optional log-transformed metric to handle highly skewed data
df['log_likes'] = np.log1p(df['likes']) # log1p safely handles 0 values
```

```python
# Groupby summaries
if 'sentiment' in df.columns:
    print("\nSummary by Sentiment:\n", df.groupby('sentiment')['engagement_score'].mean())
```

```python
# Compute stats for a specific metric (e.g., likes)
metric = 'likes'

print(f"--- Statistics for {metric} ---")
print(f"Mean: {df[metric].mean()}")
print(f"Median: {df[metric].median()}")
print(f"Mode: {df[metric].mode()[0]}")
print(f"Standard Deviation: {df[metric].std()}")
print(f"Variance: {df[metric].var()}")
print(f"25th Percentile: {df[metric].quantile(0.25)}")
print(f"75th Percentile: {df[metric].quantile(0.75)}")

# Optional: Skewness and kurtosis
print(f"Skewness: {df[metric].skew()}")
print(f"Kurtosis: {df[metric].kurtosis()}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# 1. Matplotlib Scatter: likes vs impressions
plt.figure(figsize=(8,5))
plt.scatter(df['impression_count'], df['likes'], alpha=0.5)
plt.title('Likes vs Impressions')
plt.xlabel('Impressions')
plt.ylabel('Likes')
plt.show()
#Places dots on a graph using two columns. This helps you visually spot trends, such as whether higher impressions consistently lead to higher likes.
```

```python
# 2. Matplotlib Bar: posts by category
plt.figure(figsize=(8,5))
df['post_category'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Posts by Category')
plt.show()
# Automatically counts how many times each category appears (e.g., how many text posts vs. video posts) and draws a bar chart for them.
```

```python
plt.figure(figsize=(6,6))
df['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.ylabel('')
plt.show()
```

```python
# 4. Matplotlib Histogram: age
plt.figure(figsize=(8,5))
plt.hist(df['age'].dropna(), bins=15, color='purple', edgecolor='black')
plt.title('Age Distribution')
plt.show()
```

```python
# 5. Seaborn Count plot: post type
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='post_type', palette='viridis')
plt.title('Post Type Count')
plt.show()
```

```python
# 6. Seaborn Heatmap: correlation matrix
plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
# 7. Seaborn Boxplot: engagement rate
plt.figure(figsize=(8,5))
sns.boxplot(data=df, y='engagement_rate', color='lightgreen')
plt.title('Engagement Rate Boxplot')
plt.show()
```

```python
# 8. Plotly (Interactive) Scatter chart
fig = px.scatter(df, x="watch_time_sec", y="likes", color="post_type",
                 hover_data=['post_category'], title="Interactive: Watch Time vs Likes")
fig.show()
```
